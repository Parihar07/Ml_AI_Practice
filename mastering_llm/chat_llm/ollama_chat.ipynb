```python
import requests

# Ollama runs locally on port 11434
OLLAMA_BASE_URL = "http://localhost:11434/v1/chat/completions"

# Check if Ollama is running
try:
    status = requests.get("http://localhost:11434")
    print("✓ Ollama is running")
except:
    print("✗ Ollama not running. Run 'ollama serve' in terminal")
```

```python
# List available models
models_response = requests.get("http://localhost:11434/api/tags")
if models_response.status_code == 200:
    models = models_response.json().get('models', [])
    print("Available models:")
    for model in models:
        print(f"  - {model['name']}")
else:
    print("Could not fetch models")
```

```python
# Choose a model (use one from the list above)
model = "gemma3:270m"  # Change this to your installed model

headers = {"Content-Type": "application/json"}

data = {
    "model": model,
    "messages": [
        {"role": "user", "content": "Tell me a fun fact about systems programming"}
    ]
}

# Make the API call
response = requests.post(OLLAMA_BASE_URL, headers=headers, json=data)
response.json()
```

```python
# Extract just the message
response.json()["choices"][0]["message"]["content"]
```
