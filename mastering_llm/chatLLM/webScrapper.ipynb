```python
#first projec as web scraper

webContent = fetch_website_contents("https://en.wikipedia.org/wiki/Artificial_intelligence")
print(webContent)
```

```python
system_prompt = """
You are a snarkyassistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
```

```python
user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
```

```python
def message_from_web_content(web_content: str) -> list:
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + web_content}
    ]
```

```python
message_from_web_content(webContent)
```

```python
def generate_summary(system_prompt, user_prompt_prefix, web_content):
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + web_content}
    ]
    
    response = model.generate_content(messages=messages)
    return response.text
```

```python
generate_summary(system_prompt, user_prompt_prefix, webContent)
```

```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    web_content = fetch_website_contents(url)
    summary = generate_summary(system_prompt, user_prompt_prefix, web_content)
    display(Markdown(summary))
```

```python
display_summary("https://en.wikipedia.org/wiki/Artificial_intelligence")
```
